## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### What this work can say

This work can identify and measure patterns in historical content performance and evaluate whether those patterns provide useful directional signals for future performance changes. If a future prediction task is successfully constructed and validated, the resulting predictions can support the prioritization of pages for human review.

### What this work cannot claim

This work cannot establish that any observed feature causes a change in content performance. It cannot claim to predict Google's ranking decisions or future Google behavior. It also cannot claim that a page will definitely decline based on a prediction.

Before modeling, the temporal boundaries of the historical features and future target must be verified. Information from the future target window must be excluded from the feature window to prevent data leakage. Model results will describe performance under the defined dataset and validation setup and will not guarantee future business outcomes.
